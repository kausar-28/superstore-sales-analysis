# Classification experiments
Two small experiments with k-nearest-neighbours and logistic regression. Each result is compared with a **majority-class baseline** (always predicting the most common class), because accuracy alone is misleading when the classes are imbalanced.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
```

## k-NN classifier written from scratch
Each training row carries its class label in the **last column**. The distance is computed on the feature columns only.

```python
class KNN_Classifier:
    """k-nearest-neighbours classifier (Euclidean or Manhattan distance)."""

    def __init__(self, distance_metric='euclidean'):
        if distance_metric not in ('euclidean', 'manhattan'):
            raise ValueError("distance_metric must be 'euclidean' or 'manhattan'")
        self.distance_metric = distance_metric

    def get_distance_metric(self, trainp, testp):
        # trainp = features + label, testp = features only,
        # so only the first len(trainp) - 1 values are compared
        dist = 0
        for i in range(len(trainp) - 1):
            if self.distance_metric == 'euclidean':
                dist += (trainp[i] - testp[i]) ** 2
            else:
                dist += abs(trainp[i] - testp[i])
        return np.sqrt(dist) if self.distance_metric == 'euclidean' else dist

    def nearest_neighbors(self, xtrain, testd, k):
        dis_list = [(traind, self.get_distance_metric(traind, testd)) for traind in xtrain]
        dis_list.sort(key=lambda item: item[1])
        return [dis_list[j][0] for j in range(k)]

    def predict(self, xtrain, testd, k):
        neighbours = self.nearest_neighbors(xtrain, testd, k)
        labels = [row[-1] for row in neighbours]
        return statistics.mode(labels)
```

## Experiment 1: predict the product Category from Sales and customer Segment

```python
categories = df['Category'].astype('category')
print(dict(enumerate(categories.cat.categories)))

X1 = pd.get_dummies(df[['Segment']]).astype(float)   # one-hot encode Segment
X1['Sales'] = df['Sales']
X1 = X1.to_numpy()
y1 = categories.cat.codes.to_numpy()

X1_train, X1_test, y1_train, y1_test = train_test_split(
    X1, y1, test_size=0.2, stratify=y1, random_state=2)

scaler = StandardScaler().fit(X1_train)               # fit on training data only
X1_train_s = scaler.transform(X1_train)
X1_test_s = scaler.transform(X1_test)
print(X1_train_s.shape, X1_test_s.shape)
```

```python
# attach the labels as the last column of the training data
X1_train_labeled = np.column_stack((X1_train_s, y1_train))
custom_knn = KNN_Classifier(distance_metric='euclidean')

n_eval = 500    # the pure-Python k-NN is slow; raise this for a steadier estimate
custom_pred = [custom_knn.predict(X1_train_labeled, X1_test_s[i], k=5) for i in range(n_eval)]

sk_knn = KNeighborsClassifier(n_neighbors=5).fit(X1_train_s, y1_train)

baseline1 = np.bincount(y1_test).max() / len(y1_test)
print("Majority-class baseline (full test set):", round(baseline1 * 100, 2), "%")
print("scikit-learn k-NN (full test set):      ", round(sk_knn.score(X1_test_s, y1_test) * 100, 2), "%")
print(f"scikit-learn k-NN (first {n_eval} test rows):  ", round(sk_knn.score(X1_test_s[:n_eval], y1_test[:n_eval]) * 100, 2), "%")
print(f"Custom k-NN (first {n_eval} test rows):        ", round(accuracy_score(y1_test[:n_eval], custom_pred) * 100, 2), "%")
```

**Result:** the from-scratch k-NN and scikit-learn give similar accuracy, which confirms the implementation works. Both are only slightly above the majority-class baseline, so sales amount and customer segment carry very little information about the product category.

## Experiment 2: predict whether a sale is above average
Features: sub-category, region, ship mode and customer segment (one-hot encoded).

```python
df['AboveAverageSale'] = (df['Sales'] >= df['Sales'].mean()).astype(int)

X2 = pd.get_dummies(df[['Sub-Category', 'Region', 'Ship Mode', 'Segment']]).astype(float)
y2 = df['AboveAverageSale']

X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2, y2, test_size=0.2, stratify=y2, random_state=2)

baseline2 = y2_test.value_counts(normalize=True).max()
log_reg = LogisticRegression(max_iter=1000).fit(X2_train, y2_train)

print("Baseline:", round(baseline2 * 100, 2), "%")
print("Logistic regression:", round(log_reg.score(X2_test, y2_test) * 100, 2), "%")
print(classification_report(y2_test, log_reg.predict(X2_test)))
```

```python
# same model with balanced class weights: finds more above-average sales, with more false alarms
log_reg_bal = LogisticRegression(max_iter=1000, class_weight='balanced').fit(X2_train, y2_train)
print("Logistic regression (balanced):", round(log_reg_bal.score(X2_test, y2_test) * 100, 2), "%")
print(classification_report(y2_test, log_reg_bal.predict(X2_test)))
```

**Result:** logistic regression beats the majority-class baseline by a few points. It identifies below-average sales well but misses more than half of the above-average sales. Balancing the class weights finds many more of them (higher recall) but lowers precision and overall accuracy, so the better choice depends on whether a missed high-value sale or a false alarm costs more.
